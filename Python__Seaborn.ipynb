```python
import seaborn as sns
import pandas as pd
import matplotlib.pyplot as plt


# sns.set_theme(style="darkgrid") # 設定樣式為深色網格
# sns.set_palette("pastel")
```

```python
sns.set_palette("muted")  # 使用內建muted調色板
```

```python
# 散點圖 Scatter Plot
tips = sns.load_dataset("tips")

sns.scatterplot(x="total_bill", y="tip", data=tips)
plt.title("Total Bill vs Tip")
plt.show()
```

```python
#進階 散點圖
tips = sns.load_dataset("tips")

sns.scatterplot(x="total_bill", y="tip", hue="sex", style="time", data=tips)  # hue:用於按類別改變顏色, style:用於改變標記的形狀
plt.title("Scatterplot with Categories")
plt.show()
```

```python
# 直方圖 Histogram
sns.histplot(tips["total_bill"], bins=20, kde=True) #bins: 設置直方圖的區間數, kde: 是否繪製核密度估計曲線
plt.title("Distribution of Total Bill")
plt.show()
```

```python
# 箱型圖 Box Plot
sns.boxplot(x="day", y="total_bill", data=tips)
plt.title("Box Plot of Total Bill by Day")
plt.show()
```

```python
# 箱型圖 進階
sns.boxplot(x="day", y="total_bill", hue="sex", data=tips)
plt.title("Box Plot by Day and Sex")
plt.show()
```

```python
# 長條圖 Bar Plot
sns.barplot(x="day", y="total_bill", data=tips, errorbar="sd")  #x="day"：指定X軸為day
plt.title("Average Total Bill by Day")              #y="total_bill"：指定Y軸為total_bill列
plt.show()                              #data=tips：指定資料來源為已載入的 tips 資料集
                             #errorbar="sd"：表示誤差棒使用標準差來表示資料的變異性。每個柱子的頂部會顯示該類別的消費總額的標準差。
```

```python
# 熱力圖 Heatmap
# 熱力圖適合用於顯示矩陣資料，例如相關係數：
# 只選擇數值型列計算相關係數矩陣
numeric_cols = tips.select_dtypes(include=['float64', 'int64'])  # 選擇數值型列
corr = numeric_cols.corr()  # 計算相關係數矩陣

# 繪製熱力圖
sns.heatmap(corr, annot=True, cmap="coolwarm")
plt.title("Correlation Heatmap")
plt.show()
```

```python
#  配對圖 Pair Plot
sns.pairplot(tips, hue="sex", diag_kind="kde")  #hue: 按類別上色, diag_kind: 設定對角線上的圖形類型，如直方圖或 KDE
plt.show()
```

```python
# 分面圖 FacetGrid
g = sns.FacetGrid(tips, col="sex", row="time", margin_titles=True)
g.map(sns.scatterplot, "total_bill", "tip")
plt.show()
```

```python
# 折線圖 Line Plot
fmri = sns.load_dataset("fmri")
sns.lineplot(x="timepoint", y="signal", hue="event", style="region", data=fmri)
plt.title("Line Plot Example")
plt.show()
```

```python
# Seaborn 可以直接處理Pandas的DataFrame，並按列名指定數據來源
sns.histplot(data=tips, x="total_bill", hue="sex", kde=True)
plt.title("Histogram with Pandas Integration")
plt.show()
```

```python
plt.figure(figsize=(10, 6)) # Matplotlib的plt.figure()方法調整圖表大小
sns.boxplot(x="day", y="total_bill", data=tips)
plt.title("Custom Figure Size")
plt.show()
```
