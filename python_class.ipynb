```python
class person:
  def __init__(self, name, age):
    self.name = name
    self.age = age

p1 = person("Emil", 36)

print(p1.name)
print(p1.age)
```

```python
class player:
  def __init__(self, name, hp, attack):
    self.name = (name)
    self.hp = (hp)
    self.attack = (attack)

  def show_status(self):
    print(f"角色:{self.name}")
    print(f"生命值:{self.hp}")
    print(f"攻擊力:{self.attack}")

  def take_damage(self, damage):
    self.hp -= damage
    print(f"{self.name} 受到 {damage} 點傷害 ")

p1 = player("冒險者", 100, 20)
p1.show_status()
p1.take_damage(30)
p1.show_status()
```

```python
class pet:
 def __init__(self, name, age):
  self.name = (name)
  self.age = (age)

 def introduce(self):
  print(f"我的名字{self.name}")
  print(f"我今年{self.age}歲")

dog = pet("黑糖", 5)
dog.introduce()
```

```python
class player:
  def __init__(self, name, hp=100, mp=50):
    self.name = name
    self.hp = hp
    self.mp = mp

  def show_status(self):
    print(f"角色名稱：{self.name}")
    print(f"HP：{self.hp}")
    print(f"MP：{self.mp}")

player1 = player("勇者")
player2 = player("法師", 80, 120)

player1.show_status()
player2.show_status()
```
